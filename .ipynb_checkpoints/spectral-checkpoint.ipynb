```python
import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets


def legendre_and_derivatives(P, xi):
    if P == 0: return 1.0, 0.0, 0.0
    L_nm2, L_nm1 = 1.0, xi
    dL_nm2, dL_nm1 = 0.0, 1.0
    for n in range(2, P + 1):
        L_n = ((2*n - 1) * xi * L_nm1 - (n - 1) * L_nm2) / n
        dL_n = dL_nm2 + (2*n - 1) * L_nm1
        L_nm2, L_nm1 = L_nm1, L_n
        dL_nm2, dL_nm1 = dL_nm1, dL_n
    
    # Avoid division by zero at xi = 1 or -1
    denom = 1 - xi**2
    if abs(denom) < 1e-15:
        ddL_n = 0.0 # Boundary behavior
    else:
        ddL_n = (2 * xi * dL_nm1 - P * (P + 1) * L_nm1) / denom
    return L_nm1, dL_nm1, ddL_n

def setQuads(P):

    quads = np.zeros(P + 1)
    pos = np.zeros(P + 1)
    quads[0], quads[-1] = -1.0, 1.0
    
    for k in range(1, P // 2 + (P % 2)):
        xi = -np.cos(np.pi * k / P)
        
        for _ in range(50):
            L, dL, ddL = legendre_and_derivatives(P, xi)
            delta = dL / ddL
            xi -= delta
            if abs(delta) < 1e-15:
                break
                
        quads[k] = xi
        quads[P - k] = -xi
        
    return quads

def phi(P, xi, xi_node):
    if abs(xi - xi_node) < 1e-12:
        return 1.0  # The limit of the Lagrange poly at its own node is 1

    L, dlp, _ = legendre_and_derivatives(P, xi)
    l_node, _, _ = legendre_and_derivatives(P, xi_node)

    # Standard formula for GLL basis functions
    return (-(1 - xi**2) * dlp) / (P * (P + 1) * l_node * (xi - xi_node))

xi = np.linspace(-1,1, 1000)
```

# Verification polynomial definition

Interior of the element, definition of a polynomial, solution to approximate
$$
F(\xi) = \sum_{i=0}^{N_e} c_i \xi^i
$$

```python
P = 10
order_solution = P+5

def setF(order):
    coeff = np.random.random(order) - 0.5
    def ff(x):
        F = np.zeros_like(x)
        exp=0
        for c in coeff:
            F+=c*x**exp # * np.sin(0.5*exp*np.pi*x)
            exp+=1
        return F
    return ff

q = setQuads(P)
plt.figure(figsize=[12,8])
F = setF(order_solution)
plt.plot(xi, F(xi), 'r', linewidth=2)
plt.title(f"Solution to approximate (order {order_solution})")
plt.xlabel(r'$\xi$')
plt.grid()
plt.ylabel(r'$u$')
plt.show()
```

# Base/tests functions

Lagrange polynomials to approximate the solution in the element. The solution will have the form
$$
u^\delta (\xi) = \sum_{q=0}^{P} \hat u (t) \phi_p(\xi)
$$

```python
def H(xi, P, i_quad):
    q = setQuads(P)
    h = np.empty_like(xi)
    for i in range(0,np.size(xi)):
        h[i] = phi(P,xi[i],q[i_quad])
    return h

w = widgets.IntSlider(
    value=1,
    min=0,
    max=P,
    step=1,
    description='q:',
    disabled=False,
    continuous_update=True,
    orientation='horizontal',
    readout=True,
    readout_format='d'
)

def Plot(val):
    plt.figure(figsize=[12,8])
    plt.plot(xi,H(xi,P,val))
    plt.title(r"$\phi_{v}(\xi)$ (order {p})".format(v=val,p=P))
    plt.scatter(q, np.zeros_like(q), color="black")
    plt.xlabel(r"$\xi$")
    plt.show()

widgets.interact(Plot, val = w)
```

# Polynomial fitting

Use the properties of the Lagrange polynomials to approximate our solution

```python
F_approx = np.zeros_like(xi)
fig, ax = plt.subplots(1,2)
fig.set_size_inches(16,8)
ax[0].set_title('Polynomial fitting')
ax[0].plot(xi, F(xi), "k", linewidth=2)
for i in range(0, P+1):
    ax[0].plot(xi, H(xi,P,i)*F(q[i]), ":g",linewidth=0.8)
    F_approx+=H(xi,P,i)*F(q[i])
ax[0].scatter(q,np.zeros_like(q),color="k",alpha=0.2)

ax[1].set_ylim(ax[0].get_ylim())
ax[1].set_title('Exact vs. approximation')
ax[1].scatter(q,np.zeros_like(q),color="k",alpha=0.2)
ax[1].plot(xi, F(xi), "r", linewidth = 2)
ax[1].plot(xi, F_approx, ":")
plt.show()
```

# In equation terms :
$u^\delta$ is an approximation (of order $\mathbf{P}$) of our solution of order $\mathbf{P_e}$
$$
u^\delta (\xi) = F(u(\xi))
$$
Using this, we can define the residuals of this equation :
$$
R(u^\delta) = u^\delta - F(u(\xi))
$$
Remember : 
$$
u^\delta \in \mathbf{P}^P \hspace{1cm} F \in \mathbf{P}^{P_e}
$$
The residuals will be a polynomial of order $\mathbf{P}+1$

```python
plt.figure(figsize=(12,5))
plt.title("Residuals")
ResAp = F(xi)-F_approx
plt.plot(xi, ResAp)
plt.scatter(q,np.zeros_like(q),color="k")
plt.show()
```

```python
fig, ax = plt.subplots(1,2)
fig.set_size_inches(16,8)
Ntf = 1
phi_p = H(xi,P,Ntf)
ax[0].set_title('Test function {Ntf}'.format(Ntf=Ntf))
ax[0].plot(xi, F(xi), "k", linewidth=2,alpha=0.5,label="Solution")
for i in range(0, P+1):
    if (i==0):
        ax[0].plot(xi, H(xi,P,i)*F(q[i]), ":g",linewidth=0.8, alpha=0.5, label="Trial functions")
    else:
        ax[0].plot(xi, H(xi,P,i)*F(q[i]), ":g",linewidth=0.8, alpha=0.5)
    F_approx+= H(xi,P,i)*F(q[i])
    
ax[0].scatter(q,np.zeros_like(q))

ax[0].plot(xi, phi_p,"r" ,linewidth=2, label="Test function")
ax[0].legend()

norm_res = ResAp/np.max(np.abs(ResAp))
ax[1].plot(xi,norm_res,"--k",alpha=0.2, label="Normalized res")
ax[1].scatter(q, np.zeros_like(q),color="k")
ax[1].set_title('Impact on residual')
prod = ResAp*phi_p / np.max(np.abs(ResAp*phi_p))
ax[1].plot(xi,H(xi,P,Ntf),'r',linewidth=2,label="Inner product")
ax[1].plot(xi,prod,"g",linewidth=2,label="Inner product")
ax[1].fill_between(xi, prod, color='k', alpha=0.1)
ax[1].legend()

plt.show()
```

```python
def prod(xi,P,Ntf,res):
    phi_p = H(xi,P,Ntf)
    prod = res*phi_p / np.max(np.abs(res*phi_p))
    return prod

integrals = np.empty_like(q)
for i in range(0,P+1):
    integrals[i] = np.trapz(prod(xi,P,i,ResAp),xi)
    print("Value at node {i} = {ri}".format(i=i,ri=integrals[i]))
```

# Galerkin Method for PDEs
Instead of fitting our coefficients to match specific points, we are going to make the residuals **orthogonal** to our solution

```python
import numpy as np
import matplotlib.pyplot as plt

target_values = F(xi) 

# 4. Build the System Mu = b
M = np.zeros((P + 1, P + 1))
b = np.zeros(P + 1)

# Pre-calculate basis functions
phis = []
for i in range(P + 1):
    p_vals = np.array([phi(P, x, q[i]) for x in xi])
    phis.append(p_vals)

for p_idx in range(P + 1):
    # Load Vector b_p = <phi_p, F>
    b[p_idx] = np.trapz(phis[p_idx] * target_values, xi)
    for q_idx in range(P + 1):
        # Mass Matrix M_pq = <phi_p, phi_q>
        M[p_idx, q_idx] = np.trapz(phis[p_idx] * phis[q_idx], xi)

# 5. Solve for Coefficients
u_galerkin = np.linalg.solve(M, b)

# 6. Reconstruct Solution
F_galerkin = np.zeros_like(xi)
fig, ax = plt.subplots(1,2)
fig.set_size_inches(16,8)
ax[0].set_title('Galerkin Solve')
ax[0].plot(xi, F(xi), "k", linewidth=2)
for i in range(P + 1):
    ax[0].plot(xi, phis[i]*u_galerkin[i], ":g",linewidth=0.8)
    ax[0].plot(xi, H(xi,P,i)*F(q[i]), ":k",linewidth=0.8,alpha=0.5)
    F_galerkin += u_galerkin[i] * phis[i]
ax[0].scatter(q,np.zeros_like(q),color="k",alpha=0.2)


ax[1].set_ylim(ax[0].get_ylim())
ax[1].set_title('Exact vs. approximation')
ax[1].scatter(q,np.zeros_like(q),color="k",alpha=0.2)
ax[1].plot(xi, F(xi), "r", linewidth = 2)
ax[1].plot(xi, F_galerkin, ":")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.title("Residuals")
Res = F(xi)-F_galerkin
plt.plot(xi, Res)
plt.plot(xi, ResAp, '--k', alpha=0.5)
plt.scatter(q,np.zeros_like(q),color="k")
plt.show()
```

```python
fig, ax = plt.subplots(1,2)
fig.set_size_inches(16,8)
Ntf = 2
phi_p = H(xi,P,Ntf)
ax[0].set_title('Test function {Ntf}'.format(Ntf=Ntf))
ax[0].plot(xi, F(xi), "k", linewidth=2,alpha=0.5,label="Solution")
for i in range(0, P+1):
    if (i==Ntf):
        ax[0].plot(xi, u_galerkin[i] * phis[i], ":g",linewidth=1.5, alpha=1, label="Trial functions")
    else:
        ax[0].plot(xi, u_galerkin[i] * phis[i], ":g",linewidth=0.8, alpha=0.5)
    F_approx+= u_galerkin[i] * phis[i]
    
ax[0].scatter(q,np.zeros_like(q))

ax[0].plot(xi, phi_p,"r" ,linewidth=2, label="Test function")
ax[0].legend()

norm_res = Res/np.max(np.abs(Res))
ax[1].plot(xi,norm_res,"--k",alpha=0.2, label="Normalized res")
ax[1].scatter(q, np.zeros_like(q),color="k")
ax[1].set_title('Impact on residual')
prod = Res*phi_p / np.max(np.abs(Res*phi_p))
ax[1].plot(xi,H(xi,P,Ntf),'r',linewidth=2,label="Test function")
ax[1].plot(xi,prod,"g",linewidth=2,label="Inner product")
ax[1].fill_between(xi, prod, color='k', alpha=0.1)
ax[1].legend()

plt.show()
```

```python
def prod(xi,P,Ntf,res):
    phi_p = H(xi,P,Ntf)
    prod = res*phi_p / np.max(np.abs(res*phi_p))
    return prod
    

integrals = np.empty_like(q)
for i in range(0,P+1):
    integrals[i] = np.trapz(prod(xi,P,i,Res),xi)
    print("Value at node {i} = {ri}".format(i=i,ri=integrals[i]))
```

# New base test

```python
import numpy as np
import matplotlib.pyplot as plt

# Assuming P and q are defined (e.g., P=4, q=np.linspace(-1,1,P+1))
P = 10
q = np.linspace(-1, 1, P+1)
xi = np.linspace(-1, 1, 1000) # High-res grid for plotting
eps = 100
Nq = 8 # Target node to plot

print(f"dx = {q[1]-q[0]}")

def phi(r, eps):
    # RBF defined purely on radial distance 'r'
    return np.exp(-eps*r**2)
    return 1 / np.sqrt(1 + eps*(r)**2)

# 1. Assemble the dense distance matrix A
A = np.empty((P+1, P+1))
for i in range(P+1):
    for j in range(P+1):
        A[i,j] = phi(np.abs(q[i] - q[j]), eps)

# 2. Proper Algebraic Matrix Inversion
invA = np.linalg.inv(A)

# 3. Evaluate raw RBFs on the high-res physical grid xi
Phi_grid = np.empty((P+1, len(xi)))
for i in range(P+1):
    # Distance from all points in xi to node q[i]
    Phi_grid[i, :] = phi(np.abs(xi - q[i]), eps)

# 4. Map modal RBFs to Cardinal (Nodal) RBFs
Psi_grid = invA @ Phi_grid

# 5. Plot the specific cardinal function for node Nq
plt.plot(xi, Psi_grid[Nq, :], label=f'$\psi_{{{Nq}}}(x)$')

# Verify Kronecker delta property visually
plt.scatter(q, np.zeros_like(q), color='black', label='Nodes')
plt.scatter(q[Nq], 1.0, color='red', label='Collocation Point')
plt.legend()
plt.grid(True)
plt.show()
```
